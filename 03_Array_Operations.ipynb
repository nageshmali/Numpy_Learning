```python
import numpy as np
```

## Arithmetic

```python
a1 = np.array([1,2,3,4,5])
a2 = np.array([6,7,8,9,10])

add = a1+a2
sub = a1-a2
mul = a1*a2
div = a1/a2
mod = a2%a1
ddiv = a1//a2
power = a1**a2

print("Add:",add)
print("Sub:",sub)
print("Mul:",mul)
print("Div:",div)
print("Mod:",mod)
print("FloorDiv:", ddiv)
print("Power:",power)
```

# OG Matrix Multiplication

### Remember - (m * n) @ (n * p) = (m * p)


```python
a1 = np.arange(1,5).reshape(2,2)
b1 = np.arange(11,15).reshape(2,2)
print(a1 @ b1)
print("or")
print(np.dot(a1,b1))
```

# Array Stacking

## done on vectors and not on matrices

```python
a1 = np.array([1,2,3,4,5])
b1 = np.array([5,6,7,8,9])
c1 = np.array([5,6,7,8,9])

print("Vertical Stacking")
print(np.vstack((a1,b1)))
print("Horizontal Stacking")
print(np.hstack((a1,b1)))
print("Column Stacking")
print(np.column_stack((a1,b1)))


print("Vertical Stacking")
print(np.vstack((a1,b1,c1)))
print("Horizontal Stacking")
print(np.hstack((a1,b1,c1)))
print("Column Stacking")
print(np.column_stack((a1,b1,c1)))
```
