# Prove this is valid sudoku

```python
import numpy as np
```

```python
sudoku = np.array([
    [5, 3, 4, 6, 7, 8, 9, 1, 2],
    [6, 7, 2, 1, 9, 5, 3, 4, 8],
    [1, 9, 8, 3, 4, 2, 5, 6, 7],
    [8, 5, 9, 7, 6, 1, 4, 2, 3],
    [4, 2, 6, 8, 5, 3, 7, 9, 1],
    [7, 1, 3, 9, 2, 4, 8, 5, 6],
    [9, 6, 1, 5, 3, 7, 2, 8, 4],
    [2, 8, 7, 4, 1, 9, 6, 3, 5],
    [3, 4, 5, 2, 8, 6, 1, 7, 9]
])
```

```python
### Row wise sum as 45
np.sum(sudoku, axis=1)
```

```python
### Col wise sum as 45
np.sum(sudoku, axis=0)
```

```python
### sum of 9, 3*3 boxes

for i in range(0,9,3):
    for j in range(0,9,3):
        cube = sudoku[i:i+3,j:j+3]
        print(np.sum(cube))
```
