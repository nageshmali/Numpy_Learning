# Numpy Indexing and Slicing of Vectors

```python
import numpy as np
```

```python
arr = np.arange(31,41)
arr
```

```python
print(arr[2]) # Indexing is same as python list
print(arr[-1])
print(arr[6])
```

# Vector Slicing

```python
print(arr)
print(arr[0:])
print(arr[:])
print(arr[2:7])
print(arr[-1:-6:-1])
print(arr[::2])
print(arr[1::2])
```

# Numpy Indexing and Slicing for Matrices

```python
arr = np.arange(1,31).reshape(6,5)
print(arr)
```

## Indexing

```python
print(arr[0])
print()
print(arr[0,2])
print(arr[4,4])
```

## Slicing

```python
print(arr)
```

```python
print(arr[2:4, 2:4])
print("AND")
print(arr[3:, 3:])
```

```python
print(arr[:,2])
```

# Boolean Indexing

```python
arr = np.arange(1,21)
arr
```

```python
bool_index = arr % 2 == 0
bool_index
```

```python
evenNums = arr[bool_index]
evenNums
```
